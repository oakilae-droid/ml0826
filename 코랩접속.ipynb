```python
import os
from google.colab import drive, userdata

# 1. 구글 드라이브 마운트 (세션 종료 후에도 파일 유지)
drive.mount('/content/drive')

# 2. Secrets에서 토큰 불러오기
GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')

# 3. 본인 계정 및 저장소 정보 입력
GITHUB_USER = "oakilae-droid"
GITHUB_EMAIL = "oakilae@gmail.com"
REPO_NAME = "ml0826"

# 4. Git 전역 사용자 설정
!git config --global user.name "{GITHUB_USER}"
!git config --global user.email "{GITHUB_EMAIL}"
```

```python
# 작업할 구글 드라이브 디렉토리로 이동
%cd /content/drive/MyDrive/

# Secrets 토큰을 포함한 동적 URL로 Clone
!git clone https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{REPO_NAME}.git

# 복제된 프로젝트 폴더로 이동 (이후 작업은 이 폴더 안에서 진행)
%cd {REPO_NAME}
```

```python
# 반드시 프로젝트 디렉토리 내부에서 실행
%cd /content/drive/MyDrive/{REPO_NAME}

# 원격 저장소 URL에 토큰 적용 (토큰 세션 갱신 목적)
!git remote set-url origin https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{REPO_NAME}.git

# 최신 변경사항 Pull
!git pull origin main
```
